```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
```

```python
def plot_auc(ax, m3_m2_list, file_pattern, label, linestyle='-', linewidth=2, marker='o', markersize=6, color=None, alpha=1.0):
    """
    繪製 AUC 曲線
    
    參數:
        ax: matplotlib axes 物件
        m3_m2_list: 質量點列表
        file_pattern: 檔案路徑模式
        label: 圖例標籤
        linestyle: 線條樣式 ('-', '--', '-.', ':', 'None')
        linewidth: 線條寬度
        marker: 標記樣式 ('o', 's', '^', 'v', 'D', '*', 'x', '+', 等)
        markersize: 標記大小
        color: 線條顏色 (None 表示自動選擇)
        alpha: 透明度 (0.0 到 1.0)
    """
    AUCs = []
    for m3, m2 in m3_m2_list:
        df = pd.read_csv(file_pattern.format(m3=m3, m2=m2))
        AUCs.append(df['AUC'])
    m3_m2_list = [f'{m3}/{m2}' for m3, m2 in m3_m2_list]
    ax.plot(m3_m2_list, AUCs, marker=marker, label=label, 
            linestyle=linestyle, linewidth=linewidth, markersize=markersize,
            color=color, alpha=alpha)
```
